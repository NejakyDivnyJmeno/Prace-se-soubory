# Soubory: čtení a zápis textu

## Základní přístup

- `open(..., mode="r", encoding="utf-8")` &ndash; otevření souboru pro čtení
- smyčka přes souboru (`for radek in soubor:`) &ndash; postupné načítání po řádcích
- `strip()` &ndash; odstranění bílých znaků (na začátku a konci) včetně konce řádku
- `soubor.close()` &ndash; ruční zavření souboru a uvolnění systémové prostředky

```python
soubor = open("./soubory/predmety.txt", mode="r", encoding="utf-8")
for radek in soubor:
    predmet = radek.strip()
    print(predmet)

soubor.close()
```

## Kontextový manager `with`

- nevýhody `close`: snadno jde zapomenout, blokuje systémové prostředky, při chybě se nemusí zavřít
- řešení: konstrukce `with`, která zavírá automaticky a bezpečně

```python
with open("./soubory/predmety.txt", mode="r", encoding="utf-8") as soubor:
    for radek in soubor:
        predmet = radek.strip()
        print(predmet)
```

- `read` vrací celý soubor (nebo zadaný počet znaků)
- výchozí `mode` je `r` (pouze pro čtení)

```python
with open("./soubory/predmety.txt", mode="r", encoding="utf-8") as soubor:
    text = soubor.read()
    print(text)
    radek = soubor.readline()
    print(radek)
```

**Mikroúloha**: Načti `predmety.txt` do seznamu `predmety`; vypiš počet i názvy, bez konců řádků.

```python
predmety = []
with open("./soubory/predmety.txt", encoding="utf-8") as soubor:
    for radek in soubor:
        predmety.append(radek.strip())

print(f"Počet předmětů: {len(predmety)}")
print(f"Předměty: {', '.join(predmety)}")
```

## Zápis do textového souboru

Dva módy:
- `w`: zápis (ve smyslu přepis) &ndash; soubor se vytvoří a otevře (pokud už existoval, tak se smaže obsah a otevře)
- `a`: přidávání &ndash; soubor se otevře, kurzor je na konci (pokud neexistuje, tak se vytvoří)

```python
batoh = ["meč", "pochodeň", "klíč"]

with open("soubory/ulozeny_batoh.txt", mode="w", encoding="utf-8") as soubor:
    for predmet in batoh:
        soubor.write(predmet + "\n")
```

```python
with open("soubory/zaznam.txt", mode="a", encoding="utf-8") as soubor:
    soubor.write("Hráč otevřel bránu.\n")
```

**Mikroúloha**: Ulož seznam `batoh` do `ulozeny_batoh.txt`. Pak přidej (nové otevření souboru) lektvar. Nakonec výsledný soubor načti do seznamu `ulozeny_batoh` a vypiš tento seznam.

```python
batoh = ["meč", "pochodeň", "klíč"]

with open("soubory/ulozeny_batoh.txt", mode="w", encoding="utf-8") as soubor:
    for predmet in batoh:
        soubor.write(predmet + "\n")

with open("soubory/ulozeny_batoh.txt", mode="a", encoding="utf-8") as soubor:
    soubor.write("lektvar\n")

with open("soubory/ulozeny_batoh.txt", mode="r", encoding="utf-8") as soubor:
    ulozeny_batoh = [radek.strip() for radek in soubor]

print(ulozeny_batoh)
```
