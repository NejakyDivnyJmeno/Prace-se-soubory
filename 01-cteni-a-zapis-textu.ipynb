# Soubory: čtení a zápis textu

## Základní přístup

- `open(..., mode="r", encoding="utf-8")` &ndash; otevření souboru pro čtení
- smyčka přes souboru (`for radek in soubor:`) &ndash; postupné načítání po řádcích
- `strip()` &ndash; odstranění bílých znaků (na začátku a konci) včetně konce řádku
- `soubor.close()` &ndash; ruční zavření souboru a uvolnění systémové prostředky

```python
soubor = open("./soubory/predmety.txt", mode="r", encoding="utf-8")
for radek in soubor:
    predmet = radek.strip()
    print(predmet)

soubor.close()
```

## Kontextový manager `with`

- nevýhody `close`: snadno jde zapomenout, blokuje systémové prostředky, při chybě se nemusí zavřít
- řešení: konstrukce `with`, která zavírá automaticky a bezpečně

```python
with open("./soubory/predmety.txt", mode="r", encoding="utf-8") as soubor:
    for radek in soubor:
        predmet = radek.strip()
        print(predmet)
```

- `read` vrací celý soubor (nebo zadaný počet znaků)
- výchozí `mode` je `r` (pouze pro čtení)

```python
with open("./soubory/predmety.txt", mode="r", encoding="utf-8") as soubor:
    text = soubor.read()
    print(text)
    radek = soubor.readline()
    print(radek)
```

**Mikroúloha**: Načti `predmety.txt` do seznamu `predmety`; vypiš počet i názvy, bez konců řádků.

```python
predmety = []
with open("./soubory/predmety.txt", encoding="utf-8") as soubor:
    for radek in soubor:
        predmety.append(radek.strip())

print(f"Počet předmětů: {len(predmety)}")
print(f"Předměty: {', '.join(predmety)}")
```
